```python
import folium
from folium import plugins
import pandas as pd

# Centro aproximado de Colombia
colombia_center = [4.5709, -74.2973]

# Crear mapa base con CartoDB (no tiene restricción de referer)
m = folium.Map(
    location=colombia_center,
    zoom_start=5,
    tiles='CartoDB positron'
)

# Función para asignar color según el valor de label
def get_color(label):
    if pd.isna(label):
        return 'gray'      # NaN = gris
    elif label == 1:
        return 'green'     # 1 = verde
    else:  # label == 0
        return 'red'       # 0 = rojo

# Agregar puntos del GeoDataFrame
for idx, row in gdf.iterrows():
    if row.geometry.geom_type == 'Point':
        cod_ref = row.get('COD_REF_EQ', 'Sin código')  # Obtener COD_REF_EQ o default
        color = get_color(row.get('label'))
        
        folium.CircleMarker(
            location=[row.geometry.y, row.geometry.x],
            radius=5,
            popup=f"<b>{cod_ref}</b><br>Label: {row.get('label')}",
            color=color,
            fill=True,
            fillColor=color,
            fillOpacity=0.7
        ).add_to(m)

# Mostrar el mapa
m
```

```python
## fuera de colombia
ids_out = ['EQ-50313-20210819-0013', '3EQ_EX', '63EQ_EX']
gdf_col = gdf[~gdf['COD_REF_EQ'].isin(ids_out)]
```
